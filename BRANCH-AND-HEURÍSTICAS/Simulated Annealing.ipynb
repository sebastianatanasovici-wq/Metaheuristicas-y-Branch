Simulated Annealing

Se inspira en el proceso físico de enfriamiento de metales.

El algoritmo comienza explorando muchas soluciones posibles y, ocasionalmente, acepta soluciones peores para evitar quedar atrapado en óptimos locales.

A medida que la temperatura disminuye, el algoritmo se vuelve más selectivo y converge hacia una buena solución.

```python
import random
import math

def simulated_annealing(
    solucion_inicial,
    generar_vecino,
    funcion_objetivo,
    temperatura_inicial=100,
    temperatura_minima=0.1,
    alpha=0.95,
    max_iter=1000
):

    # Solucion actual
    actual = solucion_inicial

    # Mejor solucion
    mejor = actual

    # Temperatura inicial
    T = temperatura_inicial

    # Mientras haya temperatura
    while T > temperatura_minima:

        # Iteraciones internas
        for _ in range(max_iter):

            # Genera vecino
            vecino = generar_vecino(actual)

            # Diferencia de costo
            delta = (
                funcion_objetivo(vecino)
                - funcion_objetivo(actual)
            )

            # Si mejora, acepta
            if delta < 0:

                actual = vecino

            else:

                # Probabilidad de aceptar
                prob = math.exp(-delta / T)

                # Acepta a veces
                if random.random() < prob:

                    actual = vecino

            # Actualiza mejor
            if (
                funcion_objetivo(actual)
                < funcion_objetivo(mejor)
            ):

                mejor = actual

        # Enfria temperatura
        T *= alpha

    # Retorna resultado
    return mejor, funcion_objetivo(mejor)
```

### EJEMPLO

```python
def generar_vecino(solucion):

    # Copia solucion
    vecino = solucion[:]

    # Elige posicion
    posicion = random.randint(0, len(vecino) - 1)

    # Cambia valor
    vecino[posicion] = random.randint(1, 9)

    return vecino


def funcion_objetivo(solucion):

    # Suma los valores
    return sum(solucion)


solucion_inicial = [9, 8, 7, 6]


mejor, costo = simulated_annealing(
    solucion_inicial,
    generar_vecino,
    funcion_objetivo,
    temperatura_inicial=100,
    temperatura_minima=0.1,
    alpha=0.95,
    max_iter=100
)

print("Mejor solucion:", mejor)
print("Costo:", costo)
```
