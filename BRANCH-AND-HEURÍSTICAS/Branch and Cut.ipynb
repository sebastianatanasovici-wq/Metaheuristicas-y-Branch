Branch and Cut

Extiende Branch and Bound agregando restricciones adicionales llamadas cuts o cutting planes.

Estas restricciones eliminan soluciones fraccionales o inválidas sin eliminar soluciones óptimas válidas.

Gracias a esto, el árbol de búsqueda se reduce y el algoritmo encuentra soluciones más rápido.

```python
from scipy.optimize import linprog

def branch_and_cut(
    c,
    A,
    b,
    generar_cut,
    es_entera
):

    # Repite hasta encontrar solucion entera
    while True:

        # Resuelve relajacion lineal
        resultado = linprog(
            c,
            A_ub=A,
            b_ub=b,
            method="highs"
        )

        # Si no hay solucion
        if not resultado.success:
            return None

        # Guarda solucion
        solucion = resultado.x

        # Verifica si es entera
        if es_entera(solucion):

            return solucion

        # Genera nuevo corte
        nueva_restriccion, nuevo_b = generar_cut(solucion)

        # Agrega el corte
        A.append(nueva_restriccion)
        b.append(nuevo_b)
```

### EJEMPLO 

```python
def es_entera(solucion):

    # Revisa enteros
    for valor in solucion:
        if abs(valor - round(valor)) > 0.01:
            return False

    return True


def generar_cut(solucion):

    # Corte simple
    nueva_restriccion = [1, 1]

    # Limite entero
    nuevo_b = int(sum(solucion))

    return nueva_restriccion, nuevo_b


c = [1, 1]

A = [
    [-1, -1]
]

b = [
    -3
]


solucion = branch_and_cut(
    c,
    A,
    b,
    generar_cut,
    es_entera
)

print("Solucion:", solucion)
```
