Tabu Search

Utiliza memoria para evitar volver a soluciones recientemente visitadas.

Mantiene una lista tabu que almacena movimientos o soluciones prohibidas temporalmente.

Esto permite explorar nuevas regiones del espacio de búsqueda y evitar ciclos repetitivos.

```python
def tabu_search(
    solucion_inicial,
    generar_vecinos,
    funcion_objetivo,
    max_iter=100,
    tabu_size=10
):

    # Solucion actual
    actual = solucion_inicial

    # Mejor solucion
    mejor = actual

    lista_tabu = []

    for _ in range(max_iter):

        # Genera vecinos
        vecinos = generar_vecinos(actual)

        vecinos_validos = [
            v for v in vecinos
            if v not in lista_tabu
        ]

        # Si no hay vecinos
        if not vecinos_validos:
            break

        # Mejor vecino
        mejor_vecino = min(
            vecinos_validos,
            key=funcion_objetivo
        )


        actual = mejor_vecino

        if (
            funcion_objetivo(actual)
            < funcion_objetivo(mejor)
        ):

            mejor = actual

        lista_tabu.append(actual)

        # Limita lista
        if len(lista_tabu) > tabu_size:

            lista_tabu.pop(0)

    return mejor, funcion_objetivo(mejor)
```

### EJEMPLO 

```python
def generar_vecinos(solucion):

    # Lista de vecinos
    vecinos = []

    # Cambia una posicion
    for i in range(len(solucion)):

        vecino = solucion[:]

        if vecino[i] > 1:
            vecino[i] -= 1

        vecinos.append(vecino)

    return vecinos


def funcion_objetivo(solucion):

    # Suma valores
    return sum(solucion)


solucion_inicial = [5, 4, 3]


mejor, costo = tabu_search(
    solucion_inicial,
    generar_vecinos,
    funcion_objetivo,
    max_iter=20,
    tabu_size=5
)

print("Mejor solucion:", mejor)
print("Costo:", costo)
```
