Ant 

Se basa en el comportamiento de las hormigas al buscar comida.

Las hormigas dejan feromonas en los caminos que recorren. Los caminos con más feromonas tienen mayor probabilidad de ser elegidos por otras hormigas.

Con el tiempo, las rutas más eficientes acumulan más feromonas y el algoritmo converge hacia soluciones de alta calidad.

```python
import random
import numpy as np

def ant_colony_optimization(
    construir_solucion,
    funcion_objetivo,
    actualizar_feromonas,
    num_hormigas=10,
    iteraciones=100
):

    mejor_solucion = None
    mejor_costo = float("inf")

    feromonas = {}

    for _ in range(iteraciones):

        soluciones = []

        for _ in range(num_hormigas):

            solucion = construir_solucion(
                feromonas
            )

            costo = funcion_objetivo(solucion)

            soluciones.append(
                (solucion, costo)
            )

            if costo < mejor_costo:

                mejor_costo = costo
                mejor_solucion = solucion

        actualizar_feromonas(
            feromonas,
            soluciones
        )

    return mejor_solucion, mejor_costo
```

### EJEMPLO

```python
def construir_solucion(feromonas):

    opciones = [1, 2, 3, 4, 5]

    for opcion in opciones:
        if opcion not in feromonas:
            feromonas[opcion] = 1

    solucion = []

    for _ in range(3):
        total = sum(feromonas.values())

        probabilidades = []

        for opcion in opciones:
            probabilidades.append(feromonas[opcion] / total)

        elegido = random.choices(
            opciones,
            weights=probabilidades
        )[0]

        solucion.append(elegido)

    return solucion


def funcion_objetivo(solucion):

    return sum(solucion)


def actualizar_feromonas(feromonas, soluciones):

    for opcion in feromonas:
        feromonas[opcion] *= 0.9

    for solucion, costo in soluciones:
        cantidad = 1 / costo

        for opcion in solucion:
            feromonas[opcion] += cantidad


mejor_solucion, mejor_costo = ant_colony_optimization(
    construir_solucion,
    funcion_objetivo,
    actualizar_feromonas,
    num_hormigas=10,
    iteraciones=50
)

print("Mejor solución:", mejor_solucion)
print("Mejor costo:", mejor_costo)
```
