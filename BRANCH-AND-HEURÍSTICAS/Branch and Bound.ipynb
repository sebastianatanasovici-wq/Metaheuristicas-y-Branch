Branch and Bound

Divide un problema grande en subproblemas más pequeños mediante ramas (branching).
Cada rama representa una posible decisión o solución parcial.

El algoritmo calcula bounds para estimar si una rama puede mejorar la mejor solución encontrada. Si una rama no puede producir una mejor solución, se elimina.

Esto permite reducir enormemente la cantidad de soluciones que deben explorarse.

```python
import heapq

class Nodo:

    def __init__(self, estado, nivel, costo, bound):

        # Estado actual
        self.estado = estado

        # Nivel del arbol
        self.nivel = nivel

        # Costo actual
        self.costo = costo

        # Limite estimado
        self.bound = bound

    def __lt__(self, other):

        # Compara por bound
        return self.bound < other.bound


def branch_and_bound(
    estado_inicial,
    generar_hijos,
    funcion_costo,
    funcion_bound,
    es_solucion
):

    # Mejor solucion
    mejor_solucion = None

    # Mejor costo inicial
    mejor_costo = float("inf")

    # Cola de nodos
    bounds = []

    # Nodo raiz
    raiz = Nodo(
        estado_inicial,
        0,
        funcion_costo(estado_inicial),
        funcion_bound(estado_inicial)
    )

    # Agrega raiz
    heapq.heappush(bounds, raiz)

    # Mientras existan nodos
    while bounds:

        # Saca mejor nodo
        nodo = heapq.heappop(bounds)

        # Poda si no mejora
        if nodo.bound >= mejor_costo:
            continue

        # Verifica solucion
        if es_solucion(nodo.estado):

            # Calcula costo
            costo = funcion_costo(nodo.estado)

            # Actualiza mejor
            if costo < mejor_costo:

                mejor_costo = costo
                mejor_solucion = nodo.estado

            continue

        # Genera hijos
        hijos = generar_hijos(nodo.estado)

        # Revisa hijos
        for hijo in hijos:

            # Crea nuevo nodo
            nuevo_nodo = Nodo(
                hijo,
                nodo.nivel + 1,
                funcion_costo(hijo),
                funcion_bound(hijo)
            )

            # Agrega si es mayor
            if nuevo_nodo.bound < mejor_costo:

                heapq.heappush(bounds, nuevo_nodo)

    
    return mejor_solucion, mejor_costo
```

### EJEMPLO

```python
numeros = [2, 3, 5, 7]
objetivo = 10


def generar_hijos(estado):
    hijos = []

    suma_actual = sum(estado)

    for numero in numeros:
        if suma_actual + numero <= objetivo:
            nuevo_estado = estado + [numero]
            hijos.append(nuevo_estado)

    return hijos


def funcion_costo(estado):
    return len(estado)


def funcion_bound(estado):
    return len(estado)


def es_solucion(estado):
    return sum(estado) == objetivo


estado_inicial = []


mejor_solucion, mejor_costo = branch_and_bound(
    estado_inicial,
    generar_hijos,
    funcion_costo,
    funcion_bound,
    es_solucion
)

print("Mejor solución:", mejor_solucion)
print("Mejor costo:", mejor_costo)
```
