Branch and Price

Combina Branch and Bound con generación dinámica de columnas.

En lugar de generar todas las variables del problema desde el inicio, el algoritmo crea únicamente las columnas necesarias durante la ejecución.

Esto permite resolver problemas extremadamente grandes con millones de variables potenciales.

```python
def branch_and_price(
    columnas_iniciales,
    resolver_maestro,
    pricing_problem,
    criterio_termino,
    max_iter=100
):

    # Copia columnas
    columnas = columnas_iniciales[:]

    # Itera el proceso
    for _ in range(max_iter):

        # Resuelve maestro
        solucion_maestra = resolver_maestro(columnas)

        # Busca nueva columna
        nueva_columna, costo_reducido = pricing_problem(
            columnas,
            solucion_maestra
        )

        # Verifica parada
        if criterio_termino(costo_reducido):

            break

        # Agrega columna
        columnas.append(nueva_columna)

    # Retorna columnas
    return columnas
```

### EJEMPLO

```python
def resolver_maestro(columnas):

    # Elige la menor columna
    mejor_columna = min(columnas)

    return mejor_columna


def pricing_problem(columnas, solucion_maestra):

    # Crea nueva columna
    nueva_columna = solucion_maestra - 1

    # Calcula mejora
    costo_reducido = nueva_columna

    return nueva_columna, costo_reducido


def criterio_termino(costo_reducido):

    # Termina si llega a 0
    return costo_reducido <= 0


columnas_iniciales = [5, 4, 3]


columnas_finales = branch_and_price(
    columnas_iniciales,
    resolver_maestro,
    pricing_problem,
    criterio_termino,
    max_iter=10
)

print("Columnas finales:", columnas_finales)
```
